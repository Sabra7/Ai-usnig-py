## Part 1 — The map, with coordinates this time

Same nine towns as Lab 4, but now each one has a position. That lets us compute a
**straight-line distance** to the goal, which no road can beat.

```python
import heapq, math, itertools

ROADS = {
    "Abha":            {"Khamis Mushait": 27, "Ahad Rufaidah": 45, "Al Soudah": 25},
    "Khamis Mushait":  {"Abha": 27, "Ahad Rufaidah": 22, "Tathleeth": 220, "Bisha": 330},
    "Ahad Rufaidah":   {"Abha": 45, "Khamis Mushait": 22, "Sabt Al Alaya": 185},
    "Al Soudah":       {"Abha": 25, "Rijal Almaa": 55},
    "Rijal Almaa":     {"Al Soudah": 55, "Muhayil": 60},
    "Muhayil":         {"Rijal Almaa": 60, "Sabt Al Alaya": 130},
    "Sabt Al Alaya":   {"Ahad Rufaidah": 185, "Muhayil": 130, "Bisha": 70},
    "Tathleeth":       {"Khamis Mushait": 220, "Bisha": 160},
    "Bisha":           {"Sabt Al Alaya": 70, "Tathleeth": 160, "Khamis Mushait": 330},
}

# Positions in kilometres on a flat grid, projected from real latitude/longitude.
XY = {
    "Abha": (53, 24),   "Khamis Mushait": (77, 34), "Ahad Rufaidah": (91, 22),
    "Al Soudah": (39, 31), "Rijal Almaa": (30, 21), "Muhayil": (6, 61),
    "Sabt Al Alaya": (16, 179), "Tathleeth": (160, 171), "Bisha": (63, 220),
}

class Node:
    def __init__(self, state, parent=None, action=None, cost=0):
        self.state, self.parent, self.action = state, parent, action
        self.path_cost = cost
        self.depth = 0 if parent is None else parent.depth + 1
    def path(self):
        n, out = self, []
        while n: out.append(n.state); n = n.parent
        return out[::-1]
    def __lt__(self, other): return self.path_cost < other.path_cost

class Stats:
    def __init__(self): self.expanded = 0; self.generated = 0
    def __str__(self): return f"expanded={self.expanded:<5} generated={self.generated:<5}"

print("map, coordinates and Node loaded:", len(ROADS), "towns")
```

```python
def sld(a, b):
    """Straight-line distance in km between two towns."""
    (x1, y1), (x2, y2) = XY[a], XY[b]
    return math.hypot(x1 - x2, y1 - y2)

for town in ["Abha", "Khamis Mushait", "Tathleeth", "Sabt Al Alaya"]:
    print(f"{town:<16} straight line to Bisha = {sld(town, 'Bisha'):6.1f} km")
```

### Never trust a heuristic you have not tested

A heuristic is **admissible** if it never overestimates the true remaining cost. That is the
whole condition for A\* to return an optimal solution — so check it before you rely on it.

We can check it exactly on a map this small: compute the true cheapest cost from every town to
the goal with uniform-cost search, then compare.

```python
def true_cost_to(goal):
    """Cheapest road distance from every town to goal (Dijkstra, run backwards)."""
    dist = {goal: 0}
    pq = [(0, goal)]
    while pq:
        c, u = heapq.heappop(pq)
        if c > dist.get(u, math.inf): continue
        for v, w in ROADS[u].items():
            if c + w < dist.get(v, math.inf):
                dist[v] = c + w
                heapq.heappush(pq, (c + w, v))
    return dist

def check_admissible(h, goal):
    truth = true_cost_to(goal)
    bad = [(t, round(h(t), 1), truth[t]) for t in ROADS if h(t) > truth[t] + 1e-9]
    return bad

goal = "Bisha"
h_sld = lambda t: sld(t, goal)
print("overestimates:", check_admissible(h_sld, goal) or "none - the heuristic is admissible")
```

Now break it deliberately, so you can see the checker working.

```python
h_greedy = lambda t: sld(t, goal) * 1.8      # scaled up - no longer a lower bound
print("overestimates:", check_admissible(h_greedy, goal))
```

## Part 2 — One loop, four algorithms

Lab 4 changed the *frontier* to change the algorithm. Here we keep a priority queue and change
only the **key** — the number we sort on.

| Key | Algorithm |
|---|---|
| `g(n)` | uniform-cost |
| `h(n)` | greedy best-first |
| `g(n) + h(n)` | A\* |
| `g(n) + W·h(n)`, W>1 | weighted A\* |

```python
def best_first(start, goal, key, h=None):
    """key(node) decides expansion order. h is passed in for the algorithms that use it."""
    stats = Stats()
    h = h or (lambda s: 0)
    root = Node(start)
    counter = itertools.count()                 # tie-break, keeps runs reproducible
    frontier = [(key(root, h), next(counter), root)]
    best_g = {start: 0}
    while frontier:
        _, _, node = heapq.heappop(frontier)
        if node.state == goal:
            return node, stats
        if node.path_cost > best_g.get(node.state, math.inf):
            continue                            # a cheaper route to here was already found
        stats.expanded += 1
        for nxt, w in sorted(ROADS[node.state].items()):
            g = node.path_cost + w
            if g < best_g.get(nxt, math.inf):
                best_g[nxt] = g
                child = Node(nxt, node, nxt, g)
                stats.generated += 1
                heapq.heappush(frontier, (key(child, h), next(counter), child))
    return None, stats

K_UCS    = lambda n, h: n.path_cost # g(n)
K_GREEDY = lambda n, h: h(n.state) # h(n)
K_ASTAR  = lambda n, h: n.path_cost + h(n.state) # g(n) + h(n)

def report(name, key, h=None):
    node, st = best_first("Abha", "Bisha", key, h)
    print(f"{name:<22} cost={node.path_cost:<5} {st}")
    print(f"{'':<22} {' -> '.join(node.path())}")

report("Uniform-cost",      K_UCS)
report("Greedy best-first", K_GREEDY, h_sld)
report("A*",                K_ASTAR,  h_sld)
```

### Read the three lines

- **Greedy** expands the fewest nodes — and returns the **wrong route**. It aimed straight at
  Bisha down the long desert road and never asked what that road costs.
- **Uniform-cost** expands the most. It is optimal, but it has no idea where the goal is, so it
  spreads out in every direction.
- **A\*** returns the *same optimal cost* as uniform-cost while expanding fewer nodes. That is
  the whole point of the algorithm: **optimal, and cheaper to run.**

Both informed searches used the same heuristic. The difference is entirely that A\* still counts
the kilometres already travelled, and greedy throws them away.

## Part 5 — Two heuristics for the 8-puzzle, measured

The 8-puzzle from Lab 3. Two classic heuristics:

- **h1 — misplaced tiles.** Count tiles not in their goal square.
- **h2 — Manhattan distance.** For each tile, the number of moves it needs if nothing blocked it.

Both are admissible (each move fixes at most one misplaced tile, and moves a tile one square).
**h2 dominates h1**: h2(n) ≥ h1(n) for every state. The lecture claimed that means fewer
expansions. Let us measure it.

```python
GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)          # 0 is the blank

def h1(state):
    return sum(1 for i, v in enumerate(state) if v != 0 and v != GOAL[i])

def h2(state):
    total = 0
    for i, v in enumerate(state):
        if v == 0: continue
        g = GOAL.index(v)
        total += abs(i // 3 - g // 3) + abs(i % 3 - g % 3)
    return total

def moves(state):
    b = state.index(0)
    r, c = divmod(b, 3)
    for dr, dc, name in [(-1,0,"Up"), (1,0,"Down"), (0,-1,"Left"), (0,1,"Right")]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            s = list(state); s[b], s[j] = s[j], s[b]
            yield name, tuple(s)

def astar_puzzle(start, h):
    stats = Stats()
    counter = itertools.count()
    root = Node(start)
    frontier = [(h(start), next(counter), root)]
    best_g = {start: 0}
    while frontier:
        _, _, node = heapq.heappop(frontier)
        if node.state == GOAL:
            return node, stats
        if node.path_cost > best_g.get(node.state, math.inf): continue
        stats.expanded += 1
        for name, s in moves(node.state):
            g = node.path_cost + 1
            if g < best_g.get(s, math.inf):
                best_g[s] = g
                stats.generated += 1
                heapq.heappush(frontier, (g + h(s), next(counter), Node(s, node, name, g)))
    return None, stats

START = (7, 2, 4, 5, 0, 6, 8, 3, 1)
print("h1 =", h1(START), "  h2 =", h2(START), "  (h2 >= h1, as promised)")
for name, h in [("h1 misplaced", h1), ("h2 manhattan", h2)]:
    node, st = astar_puzzle(START, h)
    print(f"{name:<14} solution length={node.depth:<3} {st}")
```

Same optimal solution length — both heuristics are admissible, so both must find it. But h2
expands far fewer nodes. **A better heuristic does not change the answer; it changes the cost of
getting there.**

And notice that h1 with A\* still beats uniform-cost, which is A\* with h = 0:

```python
node, st = astar_puzzle(START, lambda s: 0)
print(f"{'h = 0 (uniform-cost)':<22} solution length={node.depth:<3} {st}")
```

## Part 6 — Admissible is not the same as consistent

**Admissible:** h(n) never overestimates the cost to the goal.
**Consistent:** for every step n → n', h(n) ≤ cost(n, n') + h(n').

Consistency is the stronger condition, and it is what lets you *close* a state permanently — once
A\* expands a node, it has already found the cheapest route to it. With a merely admissible
heuristic you may have to reopen a closed state.

Straight-line distance is consistent, because it obeys the triangle inequality. Check it.

```python
goal = "Bisha"
bad = []
for u in ROADS:
    for v, w in ROADS[u].items():
        if sld(u, goal) > w + sld(v, goal) + 1e-9:
            bad.append((u, v))
print("consistency violations:", bad or "none")
```

### Exercise 4 — Iterative-deepening A\*

Implement IDA\*: repeated depth-first searches with a cutoff on f = g + h, where each new cutoff is the smallest f that exceeded the last one. Run it on the 8-puzzle with h2 and compare peak memory (maximum recursion depth) against A\*'s frontier size.

| | IDDFS | A* | IDA* (this code) |
|---|---|---|---|
| **Cutoff / order** | depth limit `d` | expand lowest `f` from a priority queue | DFS with `f ≤ bound`<sup>1</sup> |
| **Next bound**<sup>1</sup> | `d + 1` | none | smallest pruned `f` |
| **Uses heuristic** | no | yes | yes |
| **Optimal** | yes (unit costs) | yes (admissible `h`) | yes (admissible `h`) |
| **Memory** | O(bd) | exponential (frontier plus `best_g`) | O(bd), just the current path |
| **Time** | re-expands shallow levels | no re-expansion | re-expands earlier contours each pass |

<sup>1</sup> **Bound**: the cutoff on `f = g + h` for one depth-first pass. Any node with `f > bound` is pruned. The first bound can be `h(start)`, the most optimistic estimate of the total cost. Each later bound is the smallest `f` that was pruned in the previous pass.


```python
expanded = 0
max_depth = 0

def search(path, g, bound):
    global expanded, max_depth
    state = path[-1]
    f = g + h2(state)
    if f > bound:
        return f
    if state == GOAL:
        return "FOUND"
    expanded += 1
    max_depth = max(max_depth, len(path))
    smallest = math.inf
    for name, nxt in moves(state):
        if nxt in path:
            continue
        path.append(nxt)
        result = search(path, g + 1, bound)
        if result == "FOUND":
            return "FOUND"
        path.pop()
        smallest = min(smallest, result)
    return smallest

def ida_star(start):
    bound = h2(start)
    path = [start]
    while True:
        result = search(path, 0, bound)
        if result == "FOUND":
            return path
        bound = result

solution = ida_star(START)
print("IDA* length:", len(solution) - 1, "| expanded:", expanded, "| max depth:", max_depth)
```

```python
frontier = [(h2(START), 0, Node(START))]
best_g = {START: 0}
count = 1
peak = 1
while frontier:
    _, _, node = heapq.heappop(frontier)
    if node.state == GOAL:
        break
    if node.path_cost > best_g.get(node.state, math.inf):
        continue
    for name, s in moves(node.state):
        g = node.path_cost + 1
        if g < best_g.get(s, math.inf):
            best_g[s] = g
            heapq.heappush(frontier, (g + h2(s), count, Node(s, node, name, g)))
            count += 1
    peak = max(peak, len(frontier))
print("A* length:", node.depth, "| peak frontier:", peak)
```

A* keeps all the frontier, and its biggest frontier was 170 states (and it also keeps best_g with 452 states).
so IDA* uses much less memory, and it gets worse for A* when the puzzle is bigger.
