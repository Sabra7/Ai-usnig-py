## Part 0 — Setup: the problem definitions from Lab 3

Re-declared here so this notebook stands alone.

```python
import heapq
from collections import deque

ROADS = {
    "Abha":            {"Khamis Mushait": 27, "Ahad Rufaidah": 45, "Al Soudah": 25},
    "Khamis Mushait":  {"Abha": 27, "Ahad Rufaidah": 22, "Tathleeth": 220, "Bisha": 330},
    "Ahad Rufaidah":   {"Abha": 45, "Khamis Mushait": 22, "Sabt Al Alaya": 185},
    "Al Soudah":       {"Abha": 25, "Rijal Almaa": 55},
    "Rijal Almaa":     {"Al Soudah": 55, "Muhayil": 60},
    "Muhayil":         {"Rijal Almaa": 60, "Sabt Al Alaya": 130},
    "Sabt Al Alaya":   {"Ahad Rufaidah": 185, "Muhayil": 130, "Bisha": 70},
    "Tathleeth":       {"Khamis Mushait": 220, "Bisha": 160},
    "Bisha":           {"Sabt Al Alaya": 70, "Tathleeth": 160, "Khamis Mushait": 330},
}

class RouteProblem:
    def __init__(self, initial, goal, roads=ROADS):
        self.initial, self.goal, self.roads = initial, goal, roads
    def actions(self, s):        return sorted(self.roads[s])
    def result(self, s, a):      return a
    def goal_test(self, s):      return s == self.goal
    def step_cost(self, s, a, n): return self.roads[s][n]

print("problem ready:", RouteProblem("Abha", "Bisha").initial, "->", "Bisha")
```

## Part 1 — A node, and the loop

A **node** is a state plus the bookkeeping that lets us recover the path: its parent, the action
that produced it, and the cost so far.

Do not confuse a node with a state. Two different nodes can hold the same state, reached by
different paths at different costs.

```python
class Node:
    def __init__(self, state, parent=None, action=None, path_cost=0):
        self.state, self.parent = state, parent
        self.action, self.path_cost = action, path_cost
        self.depth = 0 if parent is None else parent.depth + 1

    def path(self):
        node, out = self, []
        while node:
            out.append(node.state)
            node = node.parent
        return list(reversed(out))

    def expand(self, problem):
        for a in problem.actions(self.state):
            nxt = problem.result(self.state, a)
            yield Node(nxt, self, a,
                       self.path_cost + problem.step_cost(self.state, a, nxt))

    def __lt__(self, other):        # so heapq can break ties without crashing
        return self.path_cost < other.path_cost

print("Node defined.")
```

### The one loop

`GRAPH-SEARCH` from the lecture. The frontier is passed in, so the *caller* decides the strategy.

```python
class Stats:
    def __init__(self): self.expanded = 0; self.generated = 0; self.max_frontier = 0
    def __repr__(self):
        return (f"expanded={self.expanded:<5} generated={self.generated:<5} "
                f"max frontier={self.max_frontier}")

def graph_search(problem, frontier, stats=None):
    """frontier must support .add(node), .pop() and .__len__()."""
    stats = stats or Stats()
    frontier.add(Node(problem.initial))
    explored = set()
    while len(frontier):
        stats.max_frontier = max(stats.max_frontier, len(frontier))
        node = frontier.pop()
        if problem.goal_test(node.state):
            return node, stats
        if node.state in explored:
            continue
        explored.add(node.state)
        stats.expanded += 1
        for child in node.expand(problem):
            stats.generated += 1
            if child.state not in explored:
                frontier.add(child)
    return None, stats

print("The loop is written once. Now we only change the frontier.")
```

## Part 2 — Four frontiers, four algorithms

This is the whole point of the lab. Read these three classes side by side.

```python
class FIFO:                     # breadth-first: shallowest node first
    def __init__(self):  self.q = deque()
    def add(self, n):    self.q.append(n)
    def pop(self):       return self.q.popleft()
    def __len__(self):   return len(self.q)

class LIFO:                     # depth-first: deepest node first
    def __init__(self):  self.q = []
    def add(self, n):    self.q.append(n)
    def pop(self):       return self.q.pop()
    def __len__(self):   return len(self.q)

class PQ:                       # uniform-cost: cheapest path so far first
    def __init__(self):  self.h = []
    def add(self, n):    heapq.heappush(self.h, (n.path_cost, n))
    def pop(self):       return heapq.heappop(self.h)[1]
    def __len__(self):   return len(self.h)

print("FIFO -> BFS      LIFO -> DFS      PQ -> uniform-cost")
```

```python
def report(name, problem, frontier):
    node, st = graph_search(problem, frontier)
    if node is None:
        print(f"{name:<22} no solution"); return
    print(f"{name:<22} cost={node.path_cost:<5} depth={node.depth}  {st}")
    print(f"{'':<22} {' -> '.join(node.path())}")

p = RouteProblem("Abha", "Bisha")
report("Breadth-first", p, FIFO())
report("Depth-first",   p, LIFO())
report("Uniform-cost",  p, PQ())
```

### Read that output carefully

**Breadth-first finds the route with the fewest towns. Uniform-cost finds the shortest one.**
They are different routes, and BFS's is longer in kilometres.

This is the lecture's claim made concrete: BFS is optimal only when every step costs the same.

## Part 3 — Iterative deepening

Depth-first search with a depth limit, run repeatedly with a larger limit each time.

```python
def depth_limited(problem, limit, stats):
    def recurse(node, depth):
        if problem.goal_test(node.state):
            return node
        if depth == 0:
            return "cutoff"
        stats.expanded += 1
        cut = False
        for child in node.expand(problem):
            stats.generated += 1
            if child.state in [n for n in child.path()[:-1]]:   # avoid cycles on this path
                continue
            r = recurse(child, depth - 1)
            if r == "cutoff":
                cut = True
            elif r is not None:
                return r
        return "cutoff" if cut else None
    r = recurse(Node(problem.initial), limit)
    return None if r in ("cutoff", None) else r

def iterative_deepening(problem, max_depth=30):
    stats = Stats()
    for limit in range(max_depth + 1):
        r = depth_limited(problem, limit, stats)
        if r is not None:
            return r, stats, limit
    return None, stats, None

node, st, limit = iterative_deepening(RouteProblem("Abha", "Bisha"))
print(f"Iterative deepening    cost={node.path_cost} depth={node.depth} found at limit {limit}")
print(f"                       {' -> '.join(node.path())}")
print(f"                       {st}")
```

### Why repeating work is cheap

IDS re-expands the shallow levels once per iteration. It sounds wasteful. Count the nodes.

```python
def bfs_nodes(b, d):   return sum(b**k for k in range(d + 1))
def ids_nodes(b, d):   return sum((d + 1 - k) * b**k for k in range(d + 1))

print(f"{'b':>3} {'d':>3} {'BFS nodes':>14} {'IDS nodes':>14} {'ratio':>8}")
for b, d in [(2, 5), (10, 3), (10, 5), (35, 4)]:
    nb_, ni = bfs_nodes(b, d), ids_nodes(b, d)
    print(f"{b:>3} {d:>3} {nb_:>14,} {ni:>14,} {ni/nb_:>8.2f}")
```

The ratio stays near 1. Almost every node in a tree lives in its deepest level, so re-walking the
shallow levels adds only a constant factor — while the memory drops from **O(bᵈ)** to **O(bd)**.

### Exercise 3 — Bidirectional search

Write a search that expands from the start and from the goal at the same time, stopping when the two frontiers meet. Compare nodes expanded against plain BFS on `Abha -> Bisha`. In a text cell, say why the saving is roughly b^(d/2) rather than b^d.

```python
def bidirectional(problem):
    start, goal = problem.initial, problem.goal
    fa, fb = deque([start]), deque([goal])
    pa, pb = {start: None}, {goal: None}
    expanded = 0

    def step(frontier, mine, other):
        nonlocal expanded
        for _ in range(len(frontier)):
            s = frontier.popleft(); expanded += 1
            for nb in problem.actions(s):
                if nb not in mine:
                    mine[nb] = s; frontier.append(nb)
                    if nb in other: return nb
        return None

    while fa and fb:
        meet = step(fa, pa, pb) or step(fb, pb, pa)
        if meet:
            left, s = [], meet
            while s is not None: left.append(s); s = pa.get(s)
            right, s = [], pb.get(meet)
            while s is not None: right.append(s); s = pb.get(s)
            return list(reversed(left)) + right, expanded
    return None, expanded

p = RouteProblem("Abha", "Bisha")
path, ex = bidirectional(p)
node, st = graph_search(p, FIFO())
print("Bidirectional:", " -> ".join(path), "| expanded =", ex)
print("Plain BFS    :", " -> ".join(node.path()), "| expanded =", st.expanded)
```

### Exercise 4 — Instrument the memory claim

Modify `graph_search` to also record the maximum size of the *explored set*. Run BFS and DFS on the 8-puzzle from Lab 3 and report both peak numbers. Does the measurement match the O(b^d) versus O(bm) claim from the lecture?

```python
def graph_search_mem(problem, frontier):
    stats = Stats()
    stats.max_explored = 0
    frontier.add(Node(problem.initial))
    explored = set()
    while len(frontier):
        stats.max_frontier = max(stats.max_frontier, len(frontier))
        stats.max_explored = max(stats.max_explored, len(explored))
        node = frontier.pop()
        if problem.goal_test(node.state):
            return node, stats
        if node.state in explored:
            continue
        explored.add(node.state)
        stats.expanded += 1
        for child in node.expand(problem):
            stats.generated += 1
            if child.state not in explored:
                frontier.add(child)
    return None, stats

class EightPuzzle:
    def __init__(self, initial, goal=(1,2,3,4,5,6,7,8,0)):
        self.initial, self.goal = initial, goal
    def actions(self, s):
        r, c = divmod(s.index(0), 3)
        return [a for a, ok in (("U", r > 0), ("D", r < 2), ("L", c > 0), ("R", c < 2)) if ok]
    def result(self, s, a):
        i = s.index(0); j = i + {"U": -3, "D": 3, "L": -1, "R": 1}[a]
        l = list(s); l[i], l[j] = l[j], l[i]
        return tuple(l)
    def goal_test(self, s):       return s == self.goal
    def step_cost(self, s, a, n): return 1

pz = EightPuzzle((1, 2, 3, 4, 0, 6, 7, 5, 8))
for name, f in [("BFS", FIFO()), ("DFS", LIFO())]:
    n, st = graph_search_mem(pz, f)
    print(f"{name}: depth={n.depth} cost={n.path_cost} {st}  max explored={st.max_explored}")
```
