## Part 1 — Sentences as data

A sentence is a nested tuple: `("&", "P", "Q")` means P ∧ Q. Symbols are plain strings.

| Connective | Written here | Meaning |
|---|---|---|
| ¬ | `("~", P)` | not |
| ∧ | `("&", P, Q)` | and |
| ∨ | `("\|", P, Q)` | or |
| ⇒ | `(">>", P, Q)` | implies |
| ⇔ | `("<=>", P, Q)` | if and only if |

```python
import itertools

def symbols(s):
    """Every propositional symbol appearing in a sentence."""
    if isinstance(s, str):
        return {s}
    out = set()
    for part in s[1:]:
        out |= symbols(part)
    return out

def evaluate(s, model):
    """Truth value of sentence s in model (a dict symbol -> True/False)."""
    if isinstance(s, str):
        return model[s]
    op = s[0]
    if op == "~":   return not evaluate(s[1], model)
    if op == "&":   return evaluate(s[1], model) and evaluate(s[2], model)
    if op == "|":   return evaluate(s[1], model) or evaluate(s[2], model)
    if op == ">>":  return (not evaluate(s[1], model)) or evaluate(s[2], model)
    if op == "<=>": return evaluate(s[1], model) == evaluate(s[2], model)
    raise ValueError(f"unknown connective: {op}")

rain = ("&", "Rain", ("~", "Umbrella"))
print("symbols:", sorted(symbols(rain)))
print("Rain=T, Umbrella=F ->", evaluate(rain, {"Rain": True, "Umbrella": False}))
print("Rain=T, Umbrella=T ->", evaluate(rain, {"Rain": True, "Umbrella": True}))
```

### The implication trap

`P ⇒ Q` is **true whenever P is false.** Students lose marks on this every year, so look at the
whole table now.

```python
imp = (">>", "P", "Q")
print(f"{'P':>6} {'Q':>6}  {'P >> Q':>8}")
for p, q in itertools.product([True, False], repeat=2):
    print(f"{str(p):>6} {str(q):>6}  {str(evaluate(imp, {'P': p, 'Q': q})):>8}")
```

"If it rains, I carry an umbrella" is not falsified by a sunny day. The only way to make an
implication false is a true antecedent with a false consequent — one row out of four.

> Implication is **not** causation, and **not** relevance. `(2+2=5) ⇒ (the moon is cheese)` is
> a true sentence. It just tells you nothing useful.

## Part 2 — Models and enumeration

A **model** assigns true or false to every symbol. With n symbols there are 2ⁿ models. Small
enough to list, for small n.

```python
def all_models(syms):
    syms = sorted(syms)
    for combo in itertools.product([True, False], repeat=len(syms)):
        yield dict(zip(syms, combo))

def truth_table(s):
    syms = sorted(symbols(s))
    head = "  ".join(f"{x:>7}" for x in syms)
    print(f"{head}  |  sentence")
    print("-" * (len(head) + 13))
    for m in all_models(syms):
        row = "  ".join(f"{str(m[x]):>7}" for x in syms)
        print(f"{row}  |  {evaluate(s, m)}")

truth_table(("<=>", ("~", ("&", "P", "Q")), ("|", ("~", "P"), ("~", "Q"))))
```

Every row is True. That sentence is **valid** — one of De Morgan's laws, proved by exhaustion.

## Part 3 — Entailment

**KB ⊨ α** means: in every model where the knowledge base is true, α is also true.

Note what that definition does *not* say. It says nothing about proof steps, rules, or search.
It is a statement about models. So we can decide it by checking them all.

```python
def entails(kb, alpha):
    """True if every model satisfying kb also satisfies alpha."""
    syms = symbols(kb) | symbols(alpha)
    for m in all_models(syms):
        if evaluate(kb, m) and not evaluate(alpha, m):
            return False, m                 # a counter-example
    return True, None

# KB: if it rains the road is wet; it is raining.
KB = ("&", (">>", "Rain", "WetRoad"), "Rain") # (Rain -> wetroad) ^ rain

for claim in ["WetRoad", "Rain", ("~", "WetRoad")]:
    ok, cex = entails(KB, claim)
    print(f"KB entails {str(claim):<22} {ok}" + (f"   counter-example: {cex}" if cex else ""))
```

### Affirming the consequent

The classic invalid inference. If the road is wet, was it raining?

```python
KB2 = ("&", (">>", "Rain", "WetRoad"), "WetRoad") # (rain or ~wetroad) ^ wetroad) =  ( false or false ^ true) = True
ok, cex = entails(KB2, "Rain")
print("KB2 entails Rain:", ok)
print("counter-example:", cex)
print("\nIn that model the road is wet without rain - a cleaning truck, say.")
print("The knowledge base never ruled it out, so the conclusion does not follow.")
```

## Part 4 — Validity and satisfiability

- **Valid** — true in every model. (A tautology.)
- **Satisfiable** — true in at least one model.
- **Unsatisfiable** — true in none.

These are the same question wearing different hats:

**KB ⊨ α  if and only if  (KB ∧ ¬α) is unsatisfiable.**

That equivalence is the foundation of every practical theorem prover, including resolution.

```python
def valid(s):
    return all(evaluate(s, m) for m in all_models(symbols(s)))

def satisfiable(s):
    for m in all_models(symbols(s)):
        if evaluate(s, m):
            return True, m
    return False, None

tests = [
    ("P | ~P            (excluded middle)", ("|", "P", ("~", "P"))),
    ("P & ~P            (contradiction)",   ("&", "P", ("~", "P"))),
    ("(P >> Q) <=> (~Q >> ~P)  (contrapositive)",
        ("<=>", (">>", "P", "Q"), (">>", ("~", "Q"), ("~", "P")))),
    ("(P >> Q) >> (Q >> P)     (converse)",
        (">>", (">>", "P", "Q"), (">>", "Q", "P"))),
]
for name, s in tests:
    sat, _ = satisfiable(s)
    print(f"{name:<44} valid={str(valid(s)):<6} satisfiable={sat}")
```

```python
# The equivalence, checked directly.
alpha = "WetRoad"
ok, _ = entails(KB, alpha)
sat, witness = satisfiable(("&", KB, ("~", alpha)))
print(f"KB entails {alpha}:              {ok}")
print(f"KB & not-{alpha} satisfiable:    {sat}")
print("The two always disagree - that is the theorem.")
```

## Part 5 — The Wumpus world

A 4×4 cave. The agent starts at [1,1]. Pits make a **breeze** in the adjacent squares. Write that
rule down for one square and the model checker can do the rest.

The rule for square [1,1]:

> B₁,₁ ⇔ (P₁,₂ ∨ P₂,₁)

*A breeze here if and only if there is a pit in one of the neighbours.* The **if and only if**
matters: plain implication would let a breeze appear from nowhere.

```python
def P(x, y): return f"P{x}{y}"
def B(x, y): return f"B{x}{y}"

def neighbours(x, y, n=4):
    return [(a, b) for a, b in [(x-1,y),(x+1,y),(x,y-1),(x,y+1)]
            if 1 <= a <= n and 1 <= b <= n]

def breeze_rule(x, y):
    nb_ = neighbours(x, y)
    disj = P(*nb_[0])
    for c in nb_[1:]:
        disj = ("|", disj, P(*c))
    return ("<=>", B(x, y), disj)

def conjoin(*parts):
    s = parts[0]
    for p in parts[1:]:
        s = ("&", s, p)
    return s

print("rule for [1,1]:", breeze_rule(1, 1))
print("rule for [2,1]:", breeze_rule(2, 1))
```

### The classic deduction

The agent has visited [1,1] (no breeze), [2,1] (breeze) and [1,2] (no breeze).
It has never seen [2,2], [3,1] or [1,3]. What can it prove?

```python
KB_W = conjoin(
    breeze_rule(1, 1), breeze_rule(2, 1), breeze_rule(1, 2),
    ("~", P(1, 1)),                       # the agent is alive, standing here
    ("~", B(1, 1)),                       # no breeze at [1,1]
    B(2, 1),                              # breeze at [2,1]
    ("~", B(1, 2)),                       # no breeze at [1,2]
)

for label, claim in [
    ("no pit at [2,2]",   ("~", P(2, 2))),
    ("no pit at [1,3]",   ("~", P(1, 3))),
    ("a pit at [3,1]",    P(3, 1)),
    ("no pit at [3,1]",   ("~", P(3, 1))),
]:
    ok, cex = entails(KB_W, claim)
    print(f"KB entails {label:<18} {ok}")
```

Read that carefully, because it is the point of the whole topic.

- **[2,2] is provably safe.** No breeze at [1,2] rules it out. The agent was never told this.
- **[1,3] is provably safe**, for the same reason.
- **[3,1] must contain a pit.** The breeze at [2,1] has to come from somewhere, and [1,1] and
  [2,2] are both eliminated. Only [3,1] is left.

The agent deduced the location of a hazard it has never been near, from three percepts and a
rule. No search, no probability — just entailment.

> This is what "knowledge-based agent" means, and it is why the course spends eight hours here.

### The cost of doing it this way

Model checking is **sound** (it never proves something false) and **complete** (it proves
everything that follows). Both properties come free from the definition of entailment.

The price is in the exponent.

```python
n = len(symbols(KB_W))
print(f"symbols in the Wumpus KB: {n}")
print(f"models enumerated:        2^{n} = {2**n:,}")
print()
for extra in [20, 40, 64]:
    print(f"with {extra} symbols: 2^{extra} = {2**extra:,} models")
```

Sixty-four propositions is a small knowledge base and already out of reach. That is the
motivation for **inference rules** — resolution, forward and backward chaining — in the next lab:
ways to reach the same conclusion without listing every world.

> **How to approach it**

> 3. **"Exactly one Wumpus"** is two parts: *at least one* (an OR over all 16 squares) and *at most one* (for every pair of squares, not both; `itertools.combinations` gives the pairs).
> 4. **Percepts.** Follow the `KB_W` cell: one line per fact, `S(1,2)` and `("~", S(1,1))`, `("~", S(2,1))`.
> 5. **Ask.** Reuse the `for ... in [...]: entails(KB, claim)` loop from Part 5, once for each square's `W(x, y)`.
> 6. **Speed.** Pits and the Wumpus are independent, so build a *separate* KB for the Wumpus. Part 5's last code cell shows why one huge KB would be far too slow.

```python
import itertools

def W(x, y): return f"W{x}{y}"
def S(x, y): return f"S{x}{y}"

def stench_rule(x, y):
    nb = neighbours(x, y)
    disj = W(*nb[0])
    for c in nb[1:]:
        disj = ("|", disj, W(*c))
    return ("<=>", S(x, y), disj)

cells = []
for x in range(1, 5):
    for y in range(1, 5):
        cells.append((x, y))

one_wumpus = W(*cells[0])
for c in cells[1:]:
    one_wumpus = ("|", one_wumpus, W(*c))

not_two = []
for a, b in itertools.combinations(cells, 2):
    not_two.append(("~", ("&", W(*a), W(*b))))
max_one = conjoin(*not_two)

KB_S = conjoin(
    stench_rule(1, 1), stench_rule(2, 1), stench_rule(1, 2),
    S(1, 2),
    ("~", S(1, 1)),
    ("~", S(2, 1)),
    one_wumpus,
    max_one,
)

for x, y in cells:
    ok, cex = entails(KB_S, W(x, y))
    print(f"wumpus at [{x},{y}]: {ok}")
```
