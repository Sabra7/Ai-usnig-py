## Part 1 — The full joint distribution

A joint distribution assigns a probability to every combination of values. Everything else in
this lab is derived from it — there is no second idea.

Our example: a student at KKU. Three variables — did they attend lectures, did they do the labs,
did they pass.

```python
import itertools
from fractions import Fraction

VARS = ["Attend", "Labs", "Pass"]

# P(Attend, Labs, Pass) - eight numbers that sum to 1
JOINT = {
    (True,  True,  True):  0.36,
    (True,  True,  False): 0.04,
    (True,  False, True):  0.12,
    (True,  False, False): 0.08,
    (False, True,  True):  0.06,
    (False, True,  False): 0.04,
    (False, False, True):  0.03,
    (False, False, False): 0.27,
}

print("number of entries:", len(JOINT), "= 2^3")
print("sums to:", round(sum(JOINT.values()), 10))
assert abs(sum(JOINT.values()) - 1) < 1e-12
```

### The three axioms

1. 0 ≤ P(a) ≤ 1
2. P(true) = 1 and P(false) = 0
3. P(a ∨ b) = P(a) + P(b) − P(a ∧ b)

Everything in probability follows from these. Check the third one directly.

```python
def prob(**evidence):
    """P of any event described by fixing some variables. Unmentioned vars are summed out."""
    total = 0.0
    for assign, p in JOINT.items():
        row = dict(zip(VARS, assign))
        if all(row[k] == v for k, v in evidence.items()):
            total += p
    return total

pa = prob(Attend=True)
pb = prob(Pass=True)
pab = prob(Attend=True, Pass=True)
print(f"P(Attend)            = {pa:.2f}")
print(f"P(Pass)              = {pb:.2f}")
print(f"P(Attend & Pass)     = {pab:.2f}")
print(f"P(Attend or Pass)    = {pa + pb - pab:.2f}   <- from axiom 3")

# verify axiom 3 by summing the rows directly
direct = sum(p for a, p in JOINT.items()
             if dict(zip(VARS, a))["Attend"] or dict(zip(VARS, a))["Pass"])
print(f"P(Attend or Pass) counted directly = {direct:.2f}")
```

## Part 2 — Marginalisation and conditioning

**Marginalisation** — to get P(X), sum the joint over everything else.

**Conditioning** — P(a | b) = P(a ∧ b) / P(b). Divide by the evidence, which renormalises the
remaining rows so they sum to 1 again.

```python
def marginal(var):
    return {v: prob(**{var: v}) for v in [True, False]}

def conditional(query_var, **evidence):
    pe = prob(**evidence)
    if pe == 0: raise ZeroDivisionError("impossible evidence")
    return {v: prob(**{query_var: v}, **evidence) / pe for v in [True, False]}

for v in VARS:
    m = marginal(v)
    print(f"P({v:<7}) = True {m[True]:.2f}   False {m[False]:.2f}")

print()
print("P(Pass | Attend=True )  =", {k: round(v, 3) for k, v in conditional("Pass", Attend=True).items()})
print("P(Pass | Attend=False)  =", {k: round(v, 3) for k, v in conditional("Pass", Attend=False).items()})
print("P(Pass | Attend=True, Labs=True) =",
      {k: round(v, 3) for k, v in conditional("Pass", Attend=True, Labs=True).items()})
```

P(Pass | Attend=True) = P(Pass ^ Attend) / P(Attend)

Attending raises the chance of passing from 0.225 to 0.80. Attending **and** doing the labs raises
it to 0.90. That is exactly the kind of question a joint distribution answers and logic cannot:
not *will they pass*, but *how much should I believe it*.

## Part 3 — Bayes' rule

$$P(\text{cause} \mid \text{effect}) = \frac{P(\text{effect} \mid \text{cause}) \; P(\text{cause})}{P(\text{effect})}$$

You almost always know P(effect | cause) — a doctor knows what a disease does — and you almost
always *want* P(cause | effect) — the patient wants to know if they have it. Bayes' rule turns
one into the other.

```python
def bayes(p_effect_given_cause, p_cause, p_effect_given_not_cause):
    p_not = 1 - p_cause
    p_effect = p_effect_given_cause * p_cause + p_effect_given_not_cause * p_not
    return p_effect_given_cause * p_cause / p_effect

# A screening test: 99% sensitive, 95% specific, for a disease affecting 1 in 1000.
sens, spec, prevalence = 0.99, 0.95, 0.001
post = bayes(sens, prevalence, 1 - spec)
print(f"test sensitivity          {sens:.0%}")
print(f"test specificity          {spec:.0%}")
print(f"prevalence                {prevalence:.1%}")
print(f"P(disease | positive)  =  {post:.1%}")
```

### Stop and look at that number

A 99%-accurate test comes back positive, and the chance you are ill is **under 2%**.

This is the **base-rate fallacy**, and it is not a trick — it is arithmetic. In a population of
100,000 with a prevalence of 1 in 1,000:

- 100 people have the disease; 99 of them test positive.
- 99,900 do not; 5% of them — 4,995 people — test positive anyway.

So 5,094 positives, of whom 99 are ill. 99 / 5094 ≈ 1.9%. The false positives swamp the true
ones because there are so many more healthy people to draw them from.

```python
N = 100_000
ill = N * prevalence
tp = ill * sens
fp = (N - ill) * (1 - spec)
print(f"{'':<22}{'positive':>10}{'negative':>10}")
print(f"{'has the disease':<22}{tp:>10.0f}{ill - tp:>10.0f}")
print(f"{'does not':<22}{fp:>10.0f}{N - ill - fp:>10.0f}")
print(f"\nP(disease | positive) = {tp:.0f} / {tp + fp:.0f} = {tp / (tp + fp):.1%}")
```

```python
# The posterior depends almost entirely on the prior. Vary it.
print(f"{'prevalence':>12} {'P(disease | positive)':>24}")
for pr in [0.0001, 0.001, 0.01, 0.10, 0.50]:
    print(f"{pr:>12.2%} {bayes(sens, pr, 1 - spec):>24.1%}")
```

> **The lesson for an AI system.** A classifier reporting 99% accuracy on a rare event is telling
> you almost nothing about what a positive prediction means. Always ask for the base rate. This
> comes back in the machine learning course as *precision on an imbalanced dataset* — same
> arithmetic, different name.

### A second test changes everything

Test again, independently, and update the posterior from the first test as your new prior.

```python
p = prevalence
for i in range(1, 4):
    p = bayes(sens, p, 1 - spec)
    print(f"after {i} positive test(s): {p:.1%}")
```

## Part 4 — Independence and conditional independence

**Independent:** P(a ∧ b) = P(a) P(b). Knowing one tells you nothing about the other.

**Conditionally independent given c:** P(a ∧ b | c) = P(a | c) P(b | c). They look related, but
only because they share a cause. Once you know the cause, the link disappears.

Conditional independence is the single most useful fact in probabilistic AI — it is what makes
Bayesian networks small enough to use.

```python
def independent(a_var, b_var, tol=1e-9):
    for av, bv in itertools.product([True, False], repeat=2):
        joint = prob(**{a_var: av, b_var: bv})
        if abs(joint - prob(**{a_var: av}) * prob(**{b_var: bv})) > tol:
            return False
    return True

print("Attend and Labs independent?", independent("Attend", "Labs"))
print("Attend and Pass  independent?", independent("Attend", "Pass"))
```

```python
# Now a distribution built so that two symptoms share one cause.
VARS2 = ["Flu", "Fever", "Ache"]
P_FLU, P_FEV, P_ACH = 0.10, 0.90, 0.80      # P(fever|flu), P(ache|flu)
P_FEV_N, P_ACH_N = 0.05, 0.10               # and given no flu

JOINT2 = {}
for f, fe, ac in itertools.product([True, False], repeat=3):
    pf = P_FLU if f else 1 - P_FLU
    pfe = (P_FEV if f else P_FEV_N) if fe else 1 - (P_FEV if f else P_FEV_N)
    pac = (P_ACH if f else P_ACH_N) if ac else 1 - (P_ACH if f else P_ACH_N)
    JOINT2[(f, fe, ac)] = pf * pfe * pac

def prob2(**ev):
    return sum(p for a, p in JOINT2.items()
               if all(dict(zip(VARS2, a))[k] == v for k, v in ev.items()))

def cond2(qv, qval, **ev):
    return prob2(**{qv: qval}, **ev) / prob2(**ev)

print(f"P(Fever)              = {prob2(Fever=True):.4f}")
print(f"P(Fever | Ache)       = {cond2('Fever', True, Ache=True):.4f}   <- they look related")
print(f"P(Fever | Flu)        = {cond2('Fever', True, Flu=True):.4f}")
print(f"P(Fever | Flu, Ache)  = {cond2('Fever', True, Flu=True, Ache=True):.4f}   <- Ache adds nothing")
```

Once you know the patient has flu, learning that they also ache does **not** change your belief
about fever. The two symptoms are conditionally independent given the disease — which is exactly
how we built the distribution, and exactly what the next lab exploits.

## Part 5 — Why the joint distribution does not scale

The joint over n boolean variables needs 2ⁿ − 1 independent numbers. Somebody has to supply them.

```python
print(f"{'variables':>10} {'joint entries':>18} {'if 1 per second':>22}")
for n in [3, 10, 20, 30, 40]:
    e = 2 ** n
    secs = e
    unit = (f"{secs/60:.0f} min" if secs < 3600 else
            f"{secs/3600:.0f} hours" if secs < 86400 * 2 else
            f"{secs/86400:.0f} days" if secs < 86400 * 800 else
            f"{secs/(86400*365):.0f} years")
    print(f"{n:>10} {e:>18,} {unit:>22}")
```

Thirty variables is a *small* problem — a medical system has hundreds — and nobody is going to
write down a billion numbers, let alone estimate them from data.

The escape is conditional independence. If Fever and Ache are independent given Flu, you do not
need the joint entry for every combination; you need P(Flu), P(Fever | Flu) and P(Ache | Flu) —
five numbers instead of seven, and the saving compounds enormously as the network grows.

That structure has a name and a set of algorithms, and it is Lab 11.

### Exercise 3 — Conditional independence in a table

Modify `JOINT2` so that Fever and Ache are *not* conditionally independent given Flu — for example, add a second shared cause. Verify with `cond2` that the independence now fails, and report how many numbers the joint needs versus the conditionally independent version.

**How to approach**

1. **Keep the original.** Make a copy, for example `JOINT3`, and leave `JOINT2` unchanged because Exercise 4 needs both. Note that `prob2` and `cond2` read the global `JOINT2`, so write versions that take the table as a parameter (e.g. `prob3`, `cond3`).
2. **Decide how to break the independence.** Two simple routes:
   - *Direct:* make the probability of Ache depend on Fever as well as Flu, i.e. define P(Ache | Flu, Fever) with a different number for each of the four (Flu, Fever) combinations.
   - *Hidden cause:* add a fourth variable (e.g. `Infection` or `Dehydrated`) that raises both Fever and Ache, build the 4-variable joint, then **sum it out** to get a joint over (Flu, Fever, Ache) only.
   Either way, the table must still sum to 1 (check it with an `assert`).
3. **Test the independence the way Part 4 does.** Compute P(Fever | Flu) and P(Fever | Flu, Ache). If conditional independence held they would be equal. They should now differ. Print both.
4. **Count the numbers needed.** For three boolean variables, the full joint needs 2³ − 1 = 7 independent numbers. The conditionally independent version (Part 5) needs 5: P(Flu), P(Fever | Flu), P(Fever | ¬Flu), P(Ache | Flu), P(Ache | ¬Flu). Count what your modified model needs when written as a product of conditional tables, and explain in a text cell what you lost by breaking independence.
5. **Watch for this trap:** if you add a hidden cause and then also *condition* on it, Fever and Ache can become independent again. The test is on the three-variable table (Flu, Fever, Ache), with the hidden cause summed out.

```python
P_ACH_GIVEN_FLU_FEV = {
    (True,  True):  0.95,
    (True,  False): 0.30,
    (False, True):  0.60,
    (False, False): 0.05,
}

JOINT3 = {}
for f, fe, ac in itertools.product([True, False], repeat=3):
    pf  = P_FLU if f else 1 - P_FLU
    pfv = P_FEV if f else P_FEV_N
    pfe = pfv if fe else 1 - pfv
    pa  = P_ACH_GIVEN_FLU_FEV[(f, fe)]
    JOINT3[(f, fe, ac)] = pf * pfe * (pa if ac else 1 - pa)

assert abs(sum(JOINT3.values()) - 1) < 1e-12
print("JOINT3 sums to:", round(sum(JOINT3.values()), 10))

def prob3(table, **ev):
    return sum(p for a, p in table.items()
               if all(dict(zip(VARS2, a))[k] == v for k, v in ev.items()))

def cond3(table, qv, qval, **ev):
    return prob3(table, **{qv: qval}, **ev) / prob3(table, **ev)

print(f"\n{'':<28}{'JOINT2':>10}{'JOINT3':>10}")
for label, kw in [("P(Fever | Flu)",        dict(Flu=True)),
                  ("P(Fever | Flu, Ache)",  dict(Flu=True, Ache=True)),
                  ("P(Fever | ¬Flu)",       dict(Flu=False)),
                  ("P(Fever | ¬Flu, Ache)", dict(Flu=False, Ache=True))]:
    print(f"{label:<28}{cond3(JOINT2, 'Fever', True, **kw):>10.4f}{cond3(JOINT3, 'Fever', True, **kw):>10.4f}")

ci = abs(cond3(JOINT3, 'Fever', True, Flu=True) - cond3(JOINT3, 'Fever', True, Flu=True, Ache=True)) < 1e-9
print("\nFever ⟂ Ache | Flu in JOINT3?", ci)

n_full = 2**3 - 1
n_ci   = 1 + 2 + 2
n_mod  = 1 + 2 + 4
print(f"\nfull joint:                 {n_full} numbers")
print(f"cond. independent (JOINT2): {n_ci} numbers")
print(f"modified model (JOINT3):    {n_mod} numbers")
```

I broke the independence by making Ache depend on both Flu and Fever, so I used P(Ache | Flu, Fever).
The table still adds up to 1.

In JOINT2, P(Fever | Flu) = P(Fever | Flu, Ache) = 0.90. In JOINT3, P(Fever | Flu) = 0.90 but
P(Fever | Flu, Ache) ≈ 0.966. Without flu it's 0.05 vs 0.387. So the independence is gone.

Number of values needed: the full joint needs 7, the independent version needs 5, and my version
needs 1 + 2 + 4 = 7. So I lost the saving and it needs as many numbers as the full joint.

### Exercise 4 — Naive Bayes from the joint

Using `JOINT2`, compute P(Flu | Fever, Ache) two ways: exactly from the joint, and with the naive Bayes approximation P(Flu) P(Fever|Flu) P(Ache|Flu) normalised. Do they agree here? Repeat after your change from Exercise 3 and explain the difference.

**How to approach**

1. **Exact answer.** Use `cond2("Flu", True, Fever=True, Ache=True)`. This conditions directly on the joint table and needs no independence assumption.
2. **Naive Bayes answer.** For each value of Flu (True and False) compute the unnormalised score P(Flu) × P(Fever=True | Flu) × P(Ache=True | Flu), taking each factor from the table (`prob2` / `cond2`) rather than retyping the constants. Then divide the Flu=True score by the sum of both scores. This normalisation plays the role of dividing by P(effect) in Bayes' rule.
3. **Compare** the two numbers. Before looking at them, predict whether they agree and why. (Hint: what did Part 4 show about how `JOINT2` was built?)
4. **Repeat on your modified table from Exercise 3.** Compute the exact answer from the modified joint, and the naive Bayes answer using the single-variable conditionals P(Fever | Flu) and P(Ache | Flu) read from that modified joint. Print all four numbers side by side in a small table.
5. **Explain (text cell):** why the two agree on `JOINT2` and not on the modified one, which one is "wrong" and in which direction (over- or under-confident), and why people still use naive Bayes on large problems.

```python
def naive_bayes_flu(table):
    score = {}
    for f in [True, False]:
        score[f] = (prob3(table, Flu=f)
                    * cond3(table, "Fever", True, Flu=f)
                    * cond3(table, "Ache",  True, Flu=f))
    return score[True] / (score[True] + score[False])

exact2 = cond2("Flu", True, Fever=True, Ache=True)
nb2    = naive_bayes_flu(JOINT2)
exact3 = cond3(JOINT3, "Flu", True, Fever=True, Ache=True)
nb3    = naive_bayes_flu(JOINT3)

print("P(Flu | Fever, Ache)")
print(f"{'table':<10}{'exact':>10}{'naive Bayes':>14}{'difference':>13}")
print(f"{'JOINT2':<10}{exact2:>10.4f}{nb2:>14.4f}{nb2 - exact2:>+13.4f}")
print(f"{'JOINT3':<10}{exact3:>10.4f}{nb3:>14.4f}{nb3 - exact3:>+13.4f}")
```

I expected them to match on JOINT2, because it was built the same way naive Bayes assumes.

| table  | exact  | naive Bayes |
|--------|--------|-------------|
| JOINT2 | 0.9412 | 0.9412      |
| JOINT3 | 0.7600 | 0.9581      |

On JOINT2 they're the same because Fever and Ache really are independent given Flu. On JOINT3 they're
different because Fever and Ache are still connected. Naive Bayes counts them as two separate pieces
of evidence, so it counts the same evidence twice. The exact answer (0.76) is the correct one, and
naive Bayes is over-confident (0.96).

People still use naive Bayes because it's simple, fast, and needs very few numbers. It usually still
picks the right class, like in spam filtering.
